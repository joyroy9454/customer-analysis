```python
# ==========================================================
# Customer Churn Data Cleaning
# Author: Joy Roy
#
# Objective:
# Clean the raw customer churn dataset by handling missing
# values, correcting invalid data, standardizing text, and
# preparing the dataset for visualization.
# ==========================================================

# Import Required Libraries
import numpy as np
import pandas as pd

# ----------------------------------------------------------
# Load Dataset
# ----------------------------------------------------------
df = pd.read_csv("customer churn.csv", encoding="latin1")

# ----------------------------------------------------------
# Initial Data Inspection
# Uncomment these lines whenever you want to inspect the data
# ----------------------------------------------------------

# print(df.info())
# print(df.describe())
# print(df.isnull().sum())
# print(df.head())

# ==========================================================
# Handling Missing and Invalid Values
# ==========================================================

# Replace missing payment methods with "Unknown"
df["Payment_Method"] = df["Payment_Method"].fillna("Unknown")

# Replace unrealistic ages
df["Age"] = np.where(df["Age"] <= 5, df["Age"].median(), df["Age"])
df["Age"] = np.where(df["Age"] > 100, df["Age"].median(), df["Age"])

# Fill missing ages with the median age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Convert age to integer
df["Age"] = df["Age"].round().astype(int)

# ==========================================================
# Cleaning Monthly Spending
# ==========================================================

# Replace extreme value with missing value
df.loc[df["Monthly_Spend"] == 99999.99, "Monthly_Spend"] = np.nan

# Check negative values
print("Negative Monthly Spend Values:",
      (df["Monthly_Spend"] < 0).sum())

# Replace negative values with missing values
df.loc[df["Monthly_Spend"] < 0, "Monthly_Spend"] = np.nan

# Fill missing values using the median
df["Monthly_Spend"] = df["Monthly_Spend"].fillna(
    df["Monthly_Spend"].median()
)

# ==========================================================
# Cleaning Customer Service Calls
# ==========================================================

# Replace negative values with missing values
df.loc[df["Customer_Service_Calls"] < 0,
       "Customer_Service_Calls"] = np.nan

# Fill missing values using the median
df["Customer_Service_Calls"] = df[
    "Customer_Service_Calls"
].fillna(df["Customer_Service_Calls"].median())

# ==========================================================
# Standardizing Text Data
# ==========================================================

df["Region"] = df["Region"].str.strip().str.title()

df["Subscription_Plan"] = (
    df["Subscription_Plan"]
      .str.strip()
      .str.title()
)

# ==========================================================
# Removing Duplicate Records
# ==========================================================

df = df.drop_duplicates()

print("Duplicate Customer IDs:",
      df["Customer_ID"].duplicated().sum())

# ==========================================================
# Optimizing Data Types
# ==========================================================

categorical_columns = [
    "Gender",
    "Region",
    "Subscription_Plan",
    "Payment_Method"
]

for column in categorical_columns:
    df[column] = df[column].astype("category")

# ==========================================================
# Final Dataset Check
# ==========================================================

print(df["Gender"].unique())
print(df["Region"].unique())
print(df["Subscription_Plan"].unique())
print(df["Payment_Method"].unique())

print(df.describe())

# ==========================================================
# Save Cleaned Dataset
# ==========================================================

df.to_csv("Customer-churn-cleaned.csv", index=False)

print("\nData cleaning completed successfully.")
print("Cleaned dataset saved as 'Customer-churn-cleaned.csv'.")
```
