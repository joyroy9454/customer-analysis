## Dataset

The dataset used in this project contains customer information from a subscription-based business. It includes demographic details, customer behavior, subscription information, and churn status.

Some important features in the dataset are:

- Age
- Gender
- Region
- Subscription Plan
- Payment Method
- Days Since Last Login
- Customer Service Calls
- Monthly Spend
- Churn

```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("Customer-churn-cleaned.csv",encoding="latin1")
```

# Section 1: Customer Profile Analysis

```python
plt.figure(figsize=(12,6))
figure = df["Age"]
plt.hist(figure,bins=5,color="Orange", edgecolor="Black")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.title("Customer Age Distribution")
plt.grid()
plt.show()

figure2 = df["Gender"].value_counts()
plt.barh(figure2.index, figure2.values, color="SkyBlue", edgecolor="Black")
plt.ylabel("Gender")
plt.xlabel("Number of Customers")
plt.title("Customer Gender Distribution")
plt.grid(axis="y")
plt.show()

figure3 = df["Region"].value_counts()
plt.bar(figure3.index, figure3.values, color="Green", edgecolor="Black")
plt.xlabel("Region")
plt.ylabel("Number of Customers")
plt.title("Regional Distribution of Customers")
plt.grid(axis="y")
plt.show()

figure4 = df["Subscription_Plan"].value_counts()
plt.bar(figure4.index, figure4.values, color="Yellow", edgecolor="Black")
plt.xlabel("Subscription Plan")
plt.ylabel("Number of Customers")
plt.title("Customer Distribution by Subscription Plan")
plt.grid(axis="y")
plt.show()

figure5 = df["Payment_Method"].value_counts()
plt.bar(figure5.index, figure5.values, color="Pink", edgecolor="Black")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.title("Customer Payment Method Distribution")
plt.grid(axis="y")
plt.show()

```

### Business Insight

The customer profile analysis shows that the dataset contains customers from different age groups, regions, subscription plans, and payment methods. Most customers are concentrated in the middle age groups, while the remaining demographic features are distributed across multiple categories. Understanding these characteristics provides a strong foundation for analyzing customer behavior and identifying factors related to churn in the following sections.

# Section 2: Customer Behavior Analysis

```python
figure1 = df["Days_Since_Last_Login"]
plt.hist(figure1,bins=5,edgecolor="Black",color="Green")
plt.xlabel("Days Since Last Login")
plt.ylabel("Number of Customers")
plt.title("Customer Login Activity")
plt.show()

figure2 = df["Customer_Service_Calls"].value_counts()
plt.bar(figure2.index,figure2.values,color="Yellow",edgecolor="Black")
plt.xlabel("Customer Service Calls")
plt.ylabel("Number of Customers")
plt.title("Distribution of Customer Service Calls")
plt.show()

figure3 = df["Monthly_Spend"]
plt.hist(figure3,bins=5,edgecolor="Black",color="Green")
plt.xlabel("Monthly Spending")
plt.ylabel("Number of Customers")
plt.title("Monthly Spending Distribution")
plt.show()

figure4 = df["Monthly_Spend"]
figure5 = df["Customer_Service_Calls"]
plt.scatter(figure4,figure5,marker="o",alpha=0.7,edgecolors="Black",color="Blue")
plt.xlabel("Monthly Spending")
plt.ylabel("Customer Service Calls")
plt.title("Monthly Spending vs Customer Service Calls")
plt.grid(True)
plt.show()

```

### Business Insight


The customer behavior analysis focuses on login activity, customer service interactions, and monthly spending patterns. These visualizations indicate that customer activity and spending vary across the dataset, while the relationship between monthly spending and customer service calls can be explored further. These behavioral features are useful for understanding customer engagement and may help explain future churn patterns.

# Section 3: Customer Churn Analysis

```python
plt.figure(figsize=(6,5))
plt.boxplot([df[df["Churn"]==0]["Age"],df[df["Churn"]==1]["Age"]],
            tick_labels=["Stayed (0)","Churned (1)"])
plt.title("Age Distribution by Churn Status")
plt.xlabel("Churn Status")
plt.ylabel("Age")
plt.show()

grouped=df.groupby(["Subscription_Plan","Churn"]).size().unstack(fill_value=0)
grouped.plot(kind="bar",figsize=(8,4))
plt.xlabel("Subscription Plan")
plt.ylabel("Number of Customers")
plt.title("Subscription Plan by Churn Status")
plt.legend(["Stayed","Churned"])
plt.grid(axis="y")
plt.show()

grouped=df.groupby(["Payment_Method","Churn"]).size().unstack(fill_value=0)
grouped.plot(kind="bar",figsize=(12,6))
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.title("Payment Method by Churn Status")
plt.legend(["Stayed","Churned"])
plt.show()

plt.boxplot([df[df["Churn"]==0]["Customer_Service_Calls"],
             df[df["Churn"]==1]["Customer_Service_Calls"]],
            tick_labels=["Stayed (0)","Churned (1)"])
plt.xlabel("Churn Status")
plt.ylabel("Customer Service Calls")
plt.title("Customer Service Calls by Churn Status")
plt.show()

plt.boxplot([df[df["Churn"]==0]["Monthly_Spend"],
             df[df["Churn"]==1]["Monthly_Spend"]],
            tick_labels=["Stayed (0)","Churned (1)"])
plt.xlabel("Churn Status")
plt.ylabel("Monthly Spending")
plt.title("Monthly Spending by Churn Status")
plt.show()

```

## Section Insight

The churn analysis compares customers who stayed with those who left the service. By examining age, subscription plans, payment methods, customer service calls, and monthly spending, we can observe differences between the two groups. These comparisons help identify characteristics that may be associated with customer churn and provide useful information for improving customer retention strategies.
